Q-Learning example

```python
import numpy as np
import random

# 6 states and 2 actions
num_states = 6
num_actions = 2

Q = np.zeros((num_states, num_actions))

alpha = 0.1
gamma = 0.9
epsilon = 0.2

episodes = 1000

goal_state = 5


for episode in range(episodes):

    state = 0

    while state != goal_state:

        # Epsilon-greedy action
        if random.random() < epsilon:
            action = random.randint(0, 1)
        else:
            action = np.argmax(Q[state])

        # Action 0 = left
        # Action 1 = right

        if action == 0:
            next_state = max(0, state - 1)

        else:
            next_state = min(goal_state, state + 1)

        # Reward
        if next_state == goal_state:
            reward = 10
        else:
            reward = -1

        # Q-learning update
        Q[state, action] = Q[state, action] + alpha * (
            reward
            + gamma * np.max(Q[next_state])
            - Q[state, action]
        )

        state = next_state


print("Learned Q-Table:\n")
print(np.round(Q, 2))

print("\nBest actions:")

for state in range(goal_state):

    action = np.argmax(Q[state])

    if action == 0:
        print("State", state, ": LEFT")
    else:
        print("State", state, ": RIGHT")
```
