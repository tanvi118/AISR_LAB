```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': ['H'],
    'F': [],
    'G': [],
    'H': []
}


# BFS
def bfs(graph, start):

    visited = set()
    queue = deque([start])

    print("BFS Traversal:")

    while queue:

        node = queue.popleft()

        if node not in visited:

            print(node, end=" ")

            visited.add(node)

            queue.extend(graph[node])


# DFS
def dfs(graph, node, visited=None):

    if visited is None:
        visited = set()

    if node not in visited:

        print(node, end=" ")

        visited.add(node)

        for neighbour in graph[node]:
            dfs(graph, neighbour, visited)


bfs(graph, 'A')

print("\n")

print("DFS Traversal:")
dfs(graph, 'A')
```
