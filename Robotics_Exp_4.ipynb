f(x) = -(x - 3)² + 10

```python
def objective(x):
    return -(x - 3)**2 + 10


def hill_climbing(start, step=0.1, iterations=100):

    current = start

    for i in range(iterations):

        left = current - step
        right = current + step

        current_value = objective(current)

        if objective(left) > current_value:
            current = left

        elif objective(right) > current_value:
            current = right

        else:
            break

    return current, objective(current)


best_x, best_value = hill_climbing(0)

print("Best x =", round(best_x, 2))
print("Maximum value =", round(best_value, 2))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 8, 200)
y = objective(x)

plt.plot(x, y)

plt.scatter(
    best_x,
    best_value,
    color="red",
    label="Maximum"
)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Hill Climbing")
plt.legend()
plt.grid()
plt.show()
```
